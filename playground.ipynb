```python
from opsrag.ingestion.discovery import discover_files
from pathlib import Path
```

```python
paths = discover_files(Path("./knowledge-base"))
paths
```

```python
from opsrag.ingestion.parsers.selector import get_parser
from opsrag.ingestion.models import IngestionResult
```

```python
ingestion_result = IngestionResult()
ingestion_result
```

```python
for file in paths:
    parser = get_parser(file)
    document = parser.parse(file, Path("./knowledge-base"))
    ingestion_result.documents.append(document)
```
